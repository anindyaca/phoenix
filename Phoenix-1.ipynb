Change Data Capture in Databricks

```python
%sql
create table default.people (first_name string, age long) using delta tblproperties (delta.enableChangeDataFeed = true)
```

```python
%sql
insert into default.people (first_name, age)
values ('Bob', 23), ('Sue', 25), ('Jim', 27)
```

```python
%sql
select * from default.people
```

```python
%sql
delete from default.people where first_name = 'Bob'
```

```python
%sql
update default.people set age = 39 where first_name = 'Jim';
    
select * from default.people
```

```python
df = spark.read.format("delta")\
    .option("readChangeFeed", "true")\
    .option("startingVersion",0)\
    .table("default.people")

display(df)
```
